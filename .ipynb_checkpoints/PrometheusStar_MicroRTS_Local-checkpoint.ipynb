## Real RTS Game with Curriculum Learning

This notebook demonstrates curriculum learning on a real-time strategy game.

### What This Does:
- Uses actual RTS game mechanics (resource gathering, unit production, combat)
- Progressive curriculum: Easy → Medium → Hard opponents
- Evolves strategies over generations
- Shows detailed game results

### Expected Runtime:
- Quick test: 5 minutes (3 games)
- Full curriculum: 2-4 hours (1000+ games)

---

```python
# Simple RTS Game Engine
import random
import time
from dataclasses import dataclass
from typing import List, Dict

@dataclass
class RTSGameState:
    """Simplified RTS game state"""
    player_resources: int = 100
    player_workers: int = 5
    player_army: int = 0
    
    opponent_resources: int = 100
    opponent_workers: int = 5
    opponent_army: int = 0
    
    turn: int = 0
    max_turns: int = 100

class RTSGame:
    """Simplified RTS game (like MicroRTS)"""
    
    def __init__(self, opponent_difficulty: int = 1):
        self.difficulty = opponent_difficulty  # 1-5
        self.state = RTSGameState()
    
    def play_game(self, strategy: Dict, verbose: bool = False) -> Dict:
        """Play a complete RTS game"""
        if verbose:
            print(f"\n🎮 Starting RTS game (opponent difficulty {self.difficulty}/5)")
        
        while self.state.turn < self.state.max_turns:
            self.state.turn += 1
            
            # Player turn (strategy-based)
            worker_ratio = strategy.get('worker_ratio', 0.3)
            army_ratio = strategy.get('army_ratio', 0.7)
            aggression = strategy.get('aggression', 0.5)
            
            # Gather resources
            self.state.player_resources += self.state.player_workers * 10
            
            # Build workers
            if self.state.player_resources >= 50 and random.random() < worker_ratio:
                self.state.player_workers += 1
                self.state.player_resources -= 50
            
            # Build army
            if self.state.player_resources >= 100 and random.random() < army_ratio:
                self.state.player_army += 2
                self.state.player_resources -= 100
            
            # Opponent turn (difficulty-based)
            self.state.opponent_resources += self.state.opponent_workers * (8 + self.difficulty * 2)
            
            if self.state.opponent_resources >= 50 and random.random() < 0.2 * self.difficulty:
                self.state.opponent_workers += 1
                self.state.opponent_resources -= 50
            
            if self.state.opponent_resources >= 100:
                self.state.opponent_army += int(1 + self.difficulty * 0.5)
                self.state.opponent_resources -= 100
            
            # Combat (if aggressive or late game)
            if self.state.turn > 20 and (random.random() < aggression or self.state.turn > 60):
                if self.state.player_army > 0 and self.state.opponent_army > 0:
                    # Battle!
                    player_losses = min(self.state.player_army, self.state.opponent_army // 2)
                    opponent_losses = min(self.state.opponent_army, self.state.player_army // 2)
                    
                    self.state.player_army -= player_losses
                    self.state.opponent_army -= opponent_losses
            
            # Check victory
            if self.state.player_army > 20 and self.state.opponent_army == 0:
                if verbose:
                    print(f"  🏆 Player wins by military (turn {self.state.turn})")
                return self._get_result(True, "Military")
            
            if self.state.opponent_army > 20 and self.state.player_army == 0:
                if verbose:
                    print(f"  💀 Opponent wins by military (turn {self.state.turn})")
                return self._get_result(False, "Military")
        
        # Time limit - score by economy
        player_score = self.state.player_resources + self.state.player_workers * 50 + self.state.player_army * 30
        opponent_score = self.state.opponent_resources + self.state.opponent_workers * 50 + self.state.opponent_army * 30
        
        won = player_score > opponent_score
        if verbose:
            result = "wins" if won else "loses"
            print(f"  {'🏆' if won else '💀'} Player {result} by economy ({player_score} vs {opponent_score})")
        
        return self._get_result(won, "Economy")
    
    def _get_result(self, won: bool, victory_type: str) -> Dict:
        return {
            'won': won,
            'victory_type': victory_type,
            'turns': self.state.turn,
            'player_army': self.state.player_army,
            'player_workers': self.state.player_workers,
            'player_resources': self.state.player_resources,
            'opponent_army': self.state.opponent_army,
            'opponent_difficulty': self.difficulty,
        }

print("✅ RTS Game Engine loaded!")
```

```python
# Test the game
print("Testing RTS game...\n")

for difficulty in [1, 3, 5]:
    strategy = {
        'worker_ratio': 0.4,
        'army_ratio': 0.6,
        'aggression': 0.5
    }
    
    game = RTSGame(opponent_difficulty=difficulty)
    result = game.play_game(strategy, verbose=True)

print("\n✅ RTS game working!")
```

## Evolutionary Algorithm

Now let's evolve strategies using a genetic algorithm.

```python
import random

class RTSEvolver:
    """Evolve RTS strategies"""
    
    def __init__(self, population_size: int = 20, mutation_rate: float = 0.3):
        self.population_size = population_size
        self.mutation_rate = mutation_rate
        self.population = []
    
    def random_strategy(self) -> Dict:
        """Create random strategy"""
        return {
            'worker_ratio': random.random(),
            'army_ratio': random.random(),
            'aggression': random.random()
        }
    
    def initialize(self):
        """Create initial population"""
        self.population = [self.random_strategy() for _ in range(self.population_size)]
    
    def evaluate(self, difficulty: int, games_per_eval: int = 3) -> List[float]:
        """Evaluate all strategies"""
        fitness_scores = []
        
        for strategy in self.population:
            wins = 0
            total_score = 0
            
            for _ in range(games_per_eval):
                game = RTSGame(opponent_difficulty=difficulty)
                result = game.play_game(strategy)
                
                if result['won']:
                    wins += 1
                
                score = (
                    (1000 if result['won'] else 0) +
                    result['player_army'] * 10 +
                    result['player_workers'] * 20 +
                    result['player_resources']
                )
                total_score += score
            
            fitness = total_score / games_per_eval
            fitness_scores.append(fitness)
        
        return fitness_scores
    
    def evolve_generation(self, difficulty: int) -> Dict:
        """Run one generation"""
        # Evaluate
        fitness = self.evaluate(difficulty)
        
        # Stats
        best_idx = fitness.index(max(fitness))
        best_strategy = self.population[best_idx]
        best_fitness = fitness[best_idx]
        avg_fitness = sum(fitness) / len(fitness)
        
        # Selection (tournament)
        parents = []
        for _ in range(self.population_size):
            tournament = random.sample(list(zip(self.population, fitness)), 3)
            winner = max(tournament, key=lambda x: x[1])[0]
            parents.append(winner)
        
        # Crossover & mutation
        new_population = [best_strategy]  # Elitism
        
        while len(new_population) < self.population_size:
            p1, p2 = random.sample(parents, 2)
            child = {}
            
            for key in p1:
                child[key] = (p1[key] + p2[key]) / 2
                
                if random.random() < self.mutation_rate:
                    child[key] = max(0, min(1, child[key] + random.gauss(0, 0.2)))
            
            new_population.append(child)
        
        self.population = new_population
        
        return {
            'best_fitness': best_fitness,
            'avg_fitness': avg_fitness,
            'best_strategy': best_strategy
        }

print("✅ Evolver ready!")
```

## Run Curriculum

Train across 3 difficulty levels with progress tracking.

```python
import time

# Curriculum stages
CURRICULUM = [
    {'stage': 1, 'name': 'Easy', 'difficulty': 1, 'generations': 50, 'target_win_rate': 0.7},
    {'stage': 2, 'name': 'Medium', 'difficulty': 3, 'generations': 75, 'target_win_rate': 0.5},
    {'stage': 3, 'name': 'Hard', 'difficulty': 5, 'generations': 100, 'target_win_rate': 0.3},
]

print("="*70)
print("🌟 PROMETHEUSSTAR RTS CURRICULUM")
print("="*70)
print()

evolver = RTSEvolver(population_size=20, mutation_rate=0.3)
evolver.initialize()

for stage in CURRICULUM:
    print(f"\n{'='*70}")
    print(f"STAGE {stage['stage']}: {stage['name']} (Difficulty {stage['difficulty']}/5)")
    print(f"Target: {stage['target_win_rate']:.0%} win rate")
    print(f"Generations: {stage['generations']}")
    print('='*70)
    print()
    
    start_time = time.time()
    best_overall = None
    
    for gen in range(stage['generations']):
        stats = evolver.evolve_generation(stage['difficulty'])
        
        if not best_overall or stats['best_fitness'] > best_overall['best_fitness']:
            best_overall = stats
        
        # Progress update every 10 gens
        if gen % 10 == 0 or gen == stage['generations'] - 1:
            # Test best strategy
            game = RTSGame(opponent_difficulty=stage['difficulty'])
            result = game.play_game(stats['best_strategy'])
            
            icon = "🏆" if result['won'] else "💀"
            elapsed = time.time() - start_time
            eta = (elapsed / (gen + 1)) * (stage['generations'] - gen - 1)
            
            print(f"Gen {gen+1:3d}/{stage['generations']} | "
                  f"Best: {stats['best_fitness']:.0f} | "
                  f"Avg: {stats['avg_fitness']:.0f} | "
                  f"{icon} {result['victory_type']} | "
                  f"Army: {result['player_army']} | "
                  f"ETA: {eta/60:.1f}m")
    
    elapsed = time.time() - start_time
    print(f"\n✅ Stage {stage['stage']} complete in {elapsed/60:.1f} minutes")
    print(f"Best fitness: {best_overall['best_fitness']:.0f}")

print("\n" + "="*70)
print("🎉 CURRICULUM COMPLETE!")
print("="*70)
```
