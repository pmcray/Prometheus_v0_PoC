## 🏗️ Part 1: Brain Map with Cell Assemblies

### What are Cell Assemblies?

Inspired by Donald Hebb's neuroscience theory, **cell assemblies** are groups of neurons (or in our case, agents) that:
- Fire together in coordinated patterns
- Form functional units for specific cognitive tasks
- Show hierarchical organization

### Prometheus's Five Core Assemblies:

1. **Planning & Strategy** (Red) - Strategic thinking and experiment design
2. **Execution & Implementation** (Green) - Code generation and action
3. **Evaluation & Analysis** (Blue) - Performance assessment
4. **Knowledge & Memory** (Orange) - Information storage and retrieval
5. **Safety & Auditing** (Purple) - Verification and compliance

```python
# Create brain map
brain = BrainMap()

print("🧠 Brain Map initialized with cell assemblies")
print("\nCore assemblies:")

stats = brain.get_stats()
for aid, details in stats['assembly_details'].items():
    print(f"  • {details['name']}: {details['members']} members")

print(f"\nTotal agents: {stats['total_nodes']}")
print(f"Total assemblies: {stats['assemblies']}")
```

```python
# Visualize initial state (all dormant)
fig = brain.visualize_assemblies(
    title="Brain Map - Cell Assemblies (Dormant State)",
    figsize=(16, 10)
)
plt.show()

print("\n💤 All assemblies dormant (0% activation)")
```

### 🔥 Simulating Activation Patterns

Let's simulate what happens when Prometheus tackles a problem:

```python
# Scenario: Planning a new experiment
print("📋 Scenario: Planning a new experiment\n")

# Activate planning assembly
brain.activate_node('PlannerAgent', level=0.9)
brain.activate_node('ExperimentOrchestrator', level=0.8)

print("Step 1: Planning assembly activated")
fig = brain.visualize_assemblies(
    title="Brain Map - Planning Phase",
    figsize=(16, 10)
)
plt.show()
```

```python
# Execution phase
print("\n⚙️ Step 2: Executing the plan\n")

brain.activate_node('CoderAgent', level=1.0)

fig = brain.visualize_assemblies(
    title="Brain Map - Execution Phase",
    figsize=(16, 10)
)
plt.show()
```

```python
# Evaluation phase
print("\n📊 Step 3: Evaluating results\n")

brain.activate_node('EvaluatorAgent', level=0.85)
brain.activate_node('ResultsSynthesizer', level=0.7)
brain.activate_node('AuditorAgent', level=0.6)  # Safety check

fig = brain.visualize_assemblies(
    title="Brain Map - Evaluation Phase",
    figsize=(16, 10)
)
plt.show()
```

```python
# Knowledge storage
print("\n💾 Step 4: Storing knowledge\n")

brain.activate_node('KnowledgeAgent', level=0.75)

# Deactivate execution (task complete)
brain.deactivate_node('CoderAgent')

fig = brain.visualize_assemblies(
    title="Brain Map - Knowledge Storage Phase",
    figsize=(16, 10)
)
plt.show()
```

```python
# Final stats
stats = brain.get_stats()

print("\n📈 Final Activation Summary:")
print(f"  Active agents: {stats['active_nodes']}/{stats['total_nodes']}\n")

for aid, details in stats['assembly_details'].items():
    activation = details['activation']
    bar = "█" * int(activation * 20)
    print(f"  {details['name']:30s} {bar:20s} {activation:5.0%}")
```

## 🔗 Part 2: Causal Agentic Mesh

### What is the Causal Agentic Mesh?

Based on **Judea Pearl's causal inference framework**, the mesh tracks:
- **Causal relationships** between agents and processes
- **Information flow** through the system
- **Critical paths** in decision-making
- **Bottlenecks** that limit performance

This enables **explainable AI** - we can answer "why did the system make this decision?"

```python
# Create example causal mesh
mesh = create_example_mesh()

print("🔗 Causal Agentic Mesh created")
print("\nThis shows a typical reasoning chain:")
print("  Observation → Perception → Planning → Decision → Action\n")

stats = mesh.get_stats()
print(f"Total nodes: {stats['total_nodes']}")
print(f"Total edges: {stats['total_edges']}")
print(f"\nNode types: {stats['node_types']}")
print(f"Edge types: {stats['edge_types']}")
```

```python
# Visualize causal mesh
fig = mesh.visualize(
    title="Causal Agentic Mesh - Example Reasoning Chain",
    figsize=(16, 12),
    highlight_active=True,
    show_edge_labels=True
)
plt.show()

print("\n🎨 Legend:")
print("  Blue = Agent")
print("  Green = Tool")
print("  Red = Decision")
print("  Orange = Observation")
print("  Gold = Active node")
print("\n  Solid line = causes")
print("  Dashed line = informs")
print("  Dotted line = depends_on")
```

### 🎯 Critical Path Analysis

Find the most important causal chain from observation to action:

```python
# Find critical path
critical_path = mesh.find_critical_path('observation_1', 'action_1')

print("🎯 Critical Path (highest causal strength):\n")
if critical_path:
    print("  " + " → ".join(critical_path))
    print(f"\n  Length: {len(critical_path)} nodes")
    
    # Calculate total strength
    total_strength = 0
    for i in range(len(critical_path) - 1):
        edge_data = mesh.graph.edges[critical_path[i], critical_path[i+1]]
        strength = edge_data.get('strength', 1.0)
        total_strength += strength
        print(f"  {critical_path[i]:20s} → {critical_path[i+1]:20s} (strength: {strength:.2f})")
    
    print(f"\n  Total causal strength: {total_strength:.2f}")
else:
    print("  No path found")
```

### 🚧 Bottleneck Detection

Identify nodes that are critical for information flow:

```python
# Find bottlenecks
bottlenecks = mesh.identify_bottlenecks(threshold=2)

print("🚧 Bottleneck Nodes:\n")
if bottlenecks:
    print("  These nodes are critical - if they fail, the system breaks!\n")
    for node in bottlenecks:
        degree = mesh.graph.degree(node)
        in_degree = mesh.graph.in_degree(node)
        out_degree = mesh.graph.out_degree(node)
        print(f"  • {node}")
        print(f"    In-degree: {in_degree}, Out-degree: {out_degree}, Total: {degree}")
else:
    print("  No bottlenecks detected")
```

### 🔍 Causal Chain Tracing

Trace all causal chains leading to a specific decision:

```python
# Get all causal chains to decision
chains = mesh.get_causal_chain('decision_1', max_depth=10)

print("🔍 All Causal Chains Leading to 'decision_1':\n")
for i, chain in enumerate(chains, 1):
    print(f"  Chain {i}: {' → '.join(chain)}")

print(f"\n  Total chains: {len(chains)}")
```

## 🎮 Part 3: Real-Time Cognitive Architecture During Game Play

Let's see how the brain map and causal mesh work together during actual game playing!

```python
# Create a simple game scenario
game = Connect4()
agent = HeuristicAgent("Red Agent", "connect4")

print("🎮 Connect 4 Game - Single Move Analysis\n")
print("Let's trace what happens in ONE move:\n")

# Create causal mesh for this move
move_mesh = CausalAgenticMesh()

# Step 1: Observe game state
move_mesh.add_node("observe_board", "observation", 1.0,
                  description="Read current board state")

# Step 2: Perception
move_mesh.add_node("identify_moves", "agent", 1.0,
                  description="Find valid moves")
move_mesh.add_edge("observe_board", "identify_moves", "causes", 1.0)

# Step 3: Analysis
move_mesh.add_node("check_win", "tool", 0.8,
                  description="Look for winning moves")
move_mesh.add_node("check_block", "tool", 0.8,
                  description="Look for blocking moves")
move_mesh.add_edge("identify_moves", "check_win", "depends_on", 0.9)
move_mesh.add_edge("identify_moves", "check_block", "depends_on", 0.9)

# Step 4: Planning
move_mesh.add_node("evaluate_options", "agent", 0.9,
                  description="Score each possible move")
move_mesh.add_edge("check_win", "evaluate_options", "informs", 0.95)
move_mesh.add_edge("check_block", "evaluate_options", "informs", 0.85)

# Step 5: Decision
move_mesh.add_node("select_move", "decision", 1.0,
                  description="Choose best move")
move_mesh.add_edge("evaluate_options", "select_move", "causes", 1.0)

# Step 6: Action
move = agent.select_move(game.get_state())
move_mesh.add_node(f"play_col_{move}", "observation", 1.0,
                  description=f"Drop token in column {move}")
move_mesh.add_edge("select_move", f"play_col_{move}", "causes", 1.0)

print(f"✓ Agent selected column {move}")
print(f"\n🧠 Agent's thinking process:")
for thought in agent.thinking:
    print(f"  • {thought}")
```

```python
# Visualize the causal chain for this single move
fig = move_mesh.visualize(
    title="Causal Mesh - Single Move Decision Process",
    figsize=(16, 12),
    highlight_active=True,
    show_edge_labels=True
)
plt.show()

print("\n💡 This shows the complete causal chain for one move!")
```

```python
# Show critical path for this move
critical = move_mesh.find_critical_path("observe_board", f"play_col_{move}")

print("🎯 Critical Path for This Move:\n")
if critical:
    for i, node in enumerate(critical):
        node_data = move_mesh.nodes[node]
        print(f"  {i+1}. {node:20s} ({node_data.node_type})")
        if 'description' in node_data.metadata:
            print(f"     → {node_data.metadata['description']}")

print(f"\n  Total steps in critical path: {len(critical)}")
```

```python
# Show the actual game board after the move
game.make_move(move)

fig = GameBoardViz.visualize_connect4(
    game.board,
    last_move=(5, move) if move is not None else None,
    title=f"Connect 4 - After Move to Column {move}"
)
plt.show()
```
