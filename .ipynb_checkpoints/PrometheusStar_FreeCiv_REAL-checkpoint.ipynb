```python
# Verify FreeCiv is installed
import subprocess

try:
    result = subprocess.run(
        ['freeciv-server', '--version'],
        capture_output=True,
        text=True,
        timeout=5
    )
    print("✅ FreeCiv Server Found!")
    print(result.stdout)
except FileNotFoundError:
    print("❌ FreeCiv not installed!")
    print("Please run: ./install_freeciv.sh")
except Exception as e:
    print(f"❌ Error: {e}")
```

## Test Real FreeCiv

Let's play a single real FreeCiv game to verify everything works.

**This will take 1-3 minutes** - you'll see actual FreeCiv server output!

```python
from freeciv_control import FreeCivGame, FreeCivConfig

print("="*70)
print("🏛️  TESTING REAL FREECIV")
print("="*70)
print()
print("Starting FreeCiv server and playing one game...")
print("This will take 1-3 minutes...")
print()

config = FreeCivConfig(
    ai_difficulty=3,
    num_civs=7,
    timeout=300,  # 5 min max
    max_turns=100  # Shorter for testing
)

game = FreeCivGame(config)
result = game.play_game(verbose=True)

print()
print("="*70)
print("✅ Real FreeCiv is working!")
print("="*70)
```

## Quick Test (Just 3 Games)

For a very quick test, let's play just 3 games at different difficulties:

```python
from freeciv_control import FreeCivGame, FreeCivConfig

print("="*70)
print("🎮 QUICK DEMO: 3 Real FreeCiv Games")
print("="*70)
print("This will take about 5-10 minutes...")
print()

for difficulty in [1, 5, 8]:
    print(f"\n{'='*70}")
    print(f"Playing at difficulty {difficulty}/10...")
    print('='*70)
    
    config = FreeCivConfig(
        ai_difficulty=difficulty,
        num_civs=7,
        timeout=300,
        max_turns=100
    )
    
    game = FreeCivGame(config)
    result = game.play_game(verbose=True)
    
    print()

print("="*70)
print("✅ Quick demo complete!")
print("="*70)
print()
print("You've seen 3 real FreeCiv games.")
print("The full curriculum plays 7,375 games like these!")
```
