## Rich Strategy Genome

Our enhanced genome includes:

### Build Priorities (weights):
- `expansion_weight` - Settlers, explorers
- `military_weight` - Units, barracks
- `economy_weight` - Markets, banks
- `science_weight` - Libraries, universities
- `culture_weight` - Temples, wonders
- `defense_weight` - Walls, fortifications

### Tech Tree:
- `tech_path` - military / science / culture / balanced

### Diplomacy:
- `diplomacy_stance` - aggressive / peaceful / neutral / isolationist
- `alliance_tendency` - 0.0 (never) to 1.0 (always)

### Military:
- `aggression_level` - 0.0 (passive) to 1.0 (very aggressive)
- `defense_ratio` - Fraction of forces for defense

### Expansion:
- `city_spacing` - Tiles between cities (3-6)
- `max_cities` - Maximum city count (6-15)

### Economy:
- `tax_rate` - Balance gold vs science (0.3-0.7)
- `luxury_rate` - Keep citizens happy (0.0-0.2)

````python
# Show example genomes
print("Example Strategy Genomes:\n")

# Military-focused genome
military_genome = StrategyGenome(
    expansion_weight=0.1,
    military_weight=0.5,
    economy_weight=0.1,
    science_weight=0.1,
    culture_weight=0.1,
    defense_weight=0.1,
    tech_path="military",
    aggression_level=0.9
)

# Science-focused genome
science_genome = StrategyGenome(
    expansion_weight=0.2,
    military_weight=0.1,
    economy_weight=0.2,
    science_weight=0.4,
    culture_weight=0.05,
    defense_weight=0.05,
    tech_path="science",
    tax_rate=0.3  # More science
)

print("Military Genome:")
print(f"  Weights: Mil={military_genome.military_weight:.1%}, Sci={military_genome.science_weight:.1%}")
print(f"  Aggression: {military_genome.aggression_level:.1%}")
print(f"  Tech Path: {military_genome.tech_path}")
print()

print("Science Genome:")
print(f"  Weights: Mil={science_genome.military_weight:.1%}, Sci={science_genome.science_weight:.1%}")
print(f"  Tax Rate: {science_genome.tax_rate:.1%} (lower = more science)")
print(f"  Tech Path: {science_genome.tech_path}")
````
