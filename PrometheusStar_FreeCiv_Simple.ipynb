- **Rich Details**: Populations, cities, technology, culture, years (4000 BC - 2050 AD)

```python
# Test the FreeCiv simulator first
from freeciv_simulator import FreeCivGame

print("="*70)
print("🎮 TESTING FREECIV SIMULATOR")
print("="*70)
print()

for difficulty in [1, 3, 5, 8]:
    print(f"Testing AI Difficulty {difficulty}/10:")
    game = FreeCivGame(ai_difficulty=difficulty)
    result = game.play_game(verbose=False)
    
    winner_icon = "🏆" if result["player_won"] else "💀"
    print(f"  {winner_icon} {result['player_civ']} vs {result['opponent_civ']}")
    print(f"     Victory: {result['victory_type']} by {result['winner']}")
    print(f"     Year: {result['final_year']} | Pop: {result['player_population']} vs {result['opponent_population']}")
    print()

print("="*70)
print("✅ FreeCiv Simulator Working!")
print("="*70)
```
