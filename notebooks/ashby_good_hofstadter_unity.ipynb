## 🕹&nbsp;[1] Programmatic Cybernetic Grid World Comparison
We will run **two parallel simulations** under the exact same environmental seed and variety disturbances:
1. **Static Baseline Agent**: Has fixed thresholds, standard metabolics, and standard double-loop properties (no causal-evidence parameter modifications).
2. **Causal-Adaptive Agent (Intelligence Explosion)**: Recursively mutates its own threshold triggers, diagonal move efficiency, and shielding absorption ratios in real-time using calculated $K(E:F)$ causal mesh weights.

```python
# @title ⚙️ Run Parallel Cybernetic Experiments
from scripts.ashby_good_hofstadter_demonstrator import (
    CyberneticGridWorld,
    AshbyHomeostaticController,
    GoodianCausalEvaluator,
    get_best_action
)

SIM_STEPS = 50

# --- RUN 1: STATIC AGENT ---
env_static = CyberneticGridWorld()
controller_static = AshbyHomeostaticController()
# Disable causal feedback updates
controller_static.apply_causal_feedback = lambda k_mesh: None

energy_static = []
safety_static = []
loop_state_static = []

for step in range(SIM_STEPS):
    controller_static.check_homeostasis()
    env_static.trigger_disturbance()
    
    energy_static.append(controller_static.energy)
    safety_static.append(controller_static.safety)
    loop_state_static.append(1.0 if controller_static.loop_state == "SECOND_LOOP" else 0.0)
    
    dx, dy = get_best_action(env_static.agent_pos, env_static, controller_static.scan_range, controller_static.diagonal_moves_unlocked)
    de, ds, _ = env_static.move_agent(dx, dy)
    controller_static.update_vitals(de, ds)

# --- RUN 2: CAUSAL-ADAPTIVE AGENT (WITH INTELLIGENCE EXPLOSION) ---
env_adaptive = CyberneticGridWorld()
controller_adaptive = AshbyHomeostaticController()
evaluator = GoodianCausalEvaluator()

energy_adaptive = []
safety_adaptive = []
loop_state_adaptive = []
disturbance_history = []
threshold_history = []
shield_history = []

print("🚀 Simulating closed-loop causal reinforcement sweep...")
for step in range(SIM_STEPS):
    controller_adaptive.check_homeostasis()
    env_adaptive.trigger_disturbance()
    
    energy_adaptive.append(controller_adaptive.energy)
    safety_adaptive.append(controller_adaptive.safety)
    loop_state_adaptive.append(1.0 if controller_adaptive.loop_state == "SECOND_LOOP" else 0.0)
    disturbance_history.append(env_adaptive.disturbance_variety)
    threshold_history.append(controller_adaptive.double_loop_threshold)
    shield_history.append((1.0 - controller_adaptive.shield_coefficient) * 100)
    
    dx, dy = get_best_action(env_adaptive.agent_pos, env_adaptive, controller_adaptive.scan_range, controller_adaptive.diagonal_moves_unlocked)
    de, ds, _ = env_adaptive.move_agent(dx, dy)
    controller_adaptive.update_vitals(de, ds)
    
    # Track and apply Goodian causal feedback every 5 steps
    success = controller_adaptive.energy > controller_adaptive.double_loop_threshold and controller_adaptive.safety > controller_adaptive.double_loop_threshold
    active_strategies = {
        "FIRST_LOOP": controller_adaptive.loop_state == "FIRST_LOOP",
        "SECOND_LOOP": controller_adaptive.loop_state == "SECOND_LOOP",
        "DIAGONALS": controller_adaptive.diagonal_moves_unlocked,
        "SHIELD": controller_adaptive.shield_active
    }
    evaluator.record_step(active_strategies, success)
    
    if step > 0 and step % 5 == 0:
        k_mesh = evaluator.get_mesh()
        controller_adaptive.apply_causal_feedback(k_mesh)

print(f"✅ Experiments complete.")
print(f"   -> Static Baseline Agent Vitals - Energy: {controller_static.energy:.1f} | Safety: {controller_static.safety:.1f}")
print(f"   -> Causal-Adaptive Agent Vitals - Energy: {controller_adaptive.energy:.1f} | Safety: {controller_adaptive.safety:.1f}")
```

## 📊 [2] Capability Ascent Curves Visualization
Let's compare the performance of both agents over the course of the 50-step run. We will visually map out:
1. **Static vs. Mutating Vitals**: View how the recursive parameter updates prevent the causal-adaptive agent from collapsing under high environmental variety disturbances.
2. **Trigger Shifting**: Watch the homeostatic threshold dynamically adapt according to Goodian causal feedback weights.

```python
# @title 📈 Plot Capability Ascent and Homeostasis
steps = np.arange(SIM_STEPS)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10), sharex=True)

# Theme Palette Colors
color_static = '#e74c3c'      # Red for static/unmutated
color_adaptive = '#2ecc71'    # Green for causal-mutating
color_safety = '#3498db'
color_disturb = '#f1c40f'

# Top Plot: Energy Homeostasis Comparison
ax1.plot(steps, energy_static, label='Static Agent Energy (Fixed 35% Trigger)', color=color_static, linewidth=2.0, linestyle='--')
ax1.plot(steps, energy_adaptive, label='Causal-Adaptive Energy (Optimizing Trigger)', color=color_adaptive, linewidth=3.0, marker='o', markersize=3)
ax1.plot(steps, threshold_history, label='Dynamic Homeostatic Threshold', color='#9b59b6', linestyle=':', linewidth=2.0)
ax1.set_ylabel('Energy Level (%)', fontsize=12, fontweight='bold')
ax1.set_title('Ashby & Good: Closed-Loop Causal Feedback vs. Static Baselines', fontsize=14, fontweight='bold', pad=15)
ax1.set_ylim(-5, 105)
ax1.legend(loc='lower left', frameon=True, facecolor='#ffffff', edgecolor='#e2e2e2')
ax1.grid(True, linestyle=':', alpha=0.6)

# Shade active second-loop cycles for the mutating agent
for s in range(SIM_STEPS):
    if loop_state_adaptive[s] == 1.0:
        ax1.axvspan(s - 0.5, s + 0.5, color='#2ecc71', alpha=0.08, label='Causal-Adaptive Second Loop' if s == 0 or (s > 0 and loop_state_adaptive[s-1] == 0.0) else "")

# Bottom Plot: Safety Vitals & Recursive Shield Optimization
ax2.plot(steps, safety_static, label='Static Agent Safety Margin', color='#95a5a6', linewidth=2.0, linestyle='--')
ax2.plot(steps, safety_adaptive, label='Causal-Adaptive Safety Margin', color=color_safety, linewidth=2.5)
ax2.plot(steps, shield_history, label='Shielding Efficiency Absorption (%)', color='#e67e22', linewidth=2.0, linestyle='-.')
ax2.set_xlabel('Simulation Steps', fontsize=12, fontweight='bold')
ax2.set_ylabel('Safety / Shielding Ratio', fontsize=12, fontweight='bold')
ax2.set_ylim(-5, 105)
ax2.legend(loc='lower left', frameon=True, facecolor='#ffffff', edgecolor='#e2e2e2')
ax2.grid(True, linestyle=':', alpha=0.6)

# Overlay environmental variety disturbance peaks
ax2_twin = ax2.twinx()
ax2_twin.fill_between(steps, disturbance_history, step='mid', label='Disturbance Variety', color=color_disturb, alpha=0.15)
ax2_twin.set_ylabel('Environmental Variety', color='#b8860b', fontsize=11, fontweight='bold')
ax2_twin.set_ylim(0, 1.0)

plt.tight_layout()
plt.show()
```

## 🧠 [3] I.J. Good: Metacognitive Causal Evidence Mesh
Here we visualize Good's Causal Weight of Evidence $K(E : F)$ computed for each active-inference strategy during the simulation run. Positively evidence-weighted strategies indicate successful survival behaviors under high environmental variety.

```python
# @title 📊 Plot Good's Causal Support Mesh K(E : F)
mesh = evaluator.get_mesh()
strategies = list(mesh.keys())
k_values = list(mesh.values())

plt.figure(figsize=(10, 5))
colors = ['#2ecc71' if k > 0 else '#e74c3c' if k < 0 else '#95a5a6' for k in k_values]

bars = plt.barh(strategies, k_values, color=colors, edgecolor='#555555', height=0.6)
plt.axvline(x=0, color='#333333', linestyle='-', linewidth=1.2)
plt.xlabel("Causal Support K(E : F) in Bits (Log-Likelihood Ratio)", fontsize=12, fontweight='bold')
plt.title("I.J. Good: Causal Weight of Evidence for Cybernetic Strategies", fontsize=13, fontweight='bold', pad=15)
plt.grid(True, linestyle=':', alpha=0.5)

# Add values to the ends of bars
for bar in bars:
    width = bar.get_width()
    label_x = width + (0.02 if width >= 0 else -0.09)
    plt.text(label_x, bar.get_y() + bar.get_height()/2, f"{width:+.4f} bits",
             va='center', ha='left' if width >= 0 else 'right', fontweight='bold', fontsize=10,
             color='#2c3e50')

plt.tight_layout()
plt.show()
```
