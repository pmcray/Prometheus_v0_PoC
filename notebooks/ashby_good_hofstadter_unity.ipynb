## 🕹️ [1] Programmatic Cybernetic Grid World Simulation
We will run the cybernetic grid world simulation programmatically to harvest fine-grained time-series metrics. The agent will navigate a dynamic survival task where environmental variety disturbances periodically spike.

```python
# @title ⚙️ Run Cybernetic Survival Experiment
from scripts.ashby_good_hofstadter_demonstrator import (
    CyberneticGridWorld,
    AshbyHomeostaticController,
    GoodianCausalEvaluator,
    get_best_action
)

# Experiment Parameters
SIM_STEPS = 50

# Initialize system
env = CyberneticGridWorld()
controller = AshbyHomeostaticController()
evaluator = GoodianCausalEvaluator()

# Metric Trackers
energy_history = []
safety_history = []
efficiency_history = []
loop_state_history = []
disturbance_history = []

print("🚀 Starting homeostatic simulation sweep...")
for step in range(SIM_STEPS):
    # Check homeostasis and trigger double-loop if needed
    controller.check_homeostasis()
    env.trigger_disturbance()
    
    # Track current metrics
    energy_history.append(controller.energy)
    safety_history.append(controller.safety)
    efficiency_history.append(controller.efficiency)
    loop_state_history.append(1.0 if controller.loop_state == "SECOND_LOOP" else 0.0)
    disturbance_history.append(env.disturbance_variety)
    
    # Select action via active inference utility mapping
    dx, dy = get_best_action(
        env.agent_pos,
        env,
        controller.scan_range,
        controller.diagonal_moves_unlocked
    )
    
    # Step environment
    de, ds, log = env.move_agent(dx, dy)
    controller.update_vitals(de, ds)
    
    # Record strategy performance for Good's Causal Mesh
    success = controller.energy > 35 and controller.safety > 35
    active_strategies = {
        "FIRST_LOOP": controller.loop_state == "FIRST_LOOP",
        "SECOND_LOOP": controller.loop_state == "SECOND_LOOP",
        "DIAGONALS": controller.diagonal_moves_unlocked,
        "SHIELD": controller.shield_active
    }
    evaluator.record_step(active_strategies, success)

print(f"✅ Simulation complete. Cycles survived: {controller.cycles_survived}/{SIM_STEPS}")
```

## 📊 [2] Ashby Homeostasis and Learning Loop Visualization
Let's plot the vital variables. We will visually map out:
1. **First-Loop vs. Second-Loop Phases**: Highlighted as shaded regions.
2. **Critical Bounds**: Red dashed threshold representing the cybernetic collapse limit.
3. **Disturbance Peaks**: Plotted alongside energy depletion to show environmental adaptation.

```python
# @title 📈 Plot Vital Variable Homeostasis
steps = np.arange(SIM_STEPS)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 9), sharex=True)

# Theme Palette Colors
color_energy = '#2ecc71'
color_safety = '#3498db'
color_critical = '#e74c3c'
color_disturb = '#f1c40f'

# Top Plot: Vital Variables
ax1.plot(steps, energy_history, label='Energy Cell Level', color=color_energy, linewidth=2.5, marker='o', markersize=4)
ax1.plot(steps, safety_history, label='Safety Margin', color=color_safety, linewidth=2.5, marker='x', markersize=4)
ax1.axhline(y=35, color=color_critical, linestyle='--', label='Critical Homeostatic Bound (35%)', linewidth=1.8)
ax1.set_ylabel('Vitals Level (%)', fontsize=12, fontweight='bold')
ax1.set_title('Ashby Cybernetic Homeostatic Engine: Double-Loop Shifting', fontsize=14, fontweight='bold', pad=15)
ax1.set_ylim(0, 105)
ax1.legend(loc='lower left', frameon=True, facecolor='#ffffff', edgecolor='#e2e2e2')
ax1.grid(True, linestyle=':', alpha=0.6)

# Shade active second-loop cycles
for s in range(SIM_STEPS):
    if loop_state_history[s] == 1.0:
        ax1.axvspan(s - 0.5, s + 0.5, color='#e74c3c', alpha=0.15, label='Second-Loop Shift' if s == 0 or (s > 0 and loop_state_history[s-1] == 0.0) else "")

# Bottom Plot: Environmental Variety Disturbance & Efficiency
ax2.fill_between(steps, disturbance_history, step='mid', label='Environmental Variety (Disturbance)', color=color_disturb, alpha=0.35)
ax2.plot(steps, efficiency_history, label='System Computational Efficiency', color='#9b59b6', linewidth=2.0)
ax2.set_xlabel('Simulation Steps', fontsize=12, fontweight='bold')
ax2.set_ylabel('Variety / Efficiency', fontsize=12, fontweight='bold')
ax2.set_ylim(0, 105)
ax2.legend(loc='upper right', frameon=True, facecolor='#ffffff', edgecolor='#e2e2e2')
ax2.grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

## 🧠 [3] I.J. Good: Metacognitive Causal Evidence Mesh
Here we visualize Good's Causal Weight of Evidence $K(E : F)$ computed for each active-inference strategy during the simulation run. Positively evidence-weighted strategies indicate successful survival behaviors under high environmental variety.

```python
# @title 📊 Plot Good's Causal Support Mesh K(E : F)
mesh = evaluator.get_mesh()
strategies = list(mesh.keys())
k_values = list(mesh.values())

plt.figure(figsize=(10, 5))
colors = ['#2ecc71' if k > 0 else '#e74c3c' if k < 0 else '#95a5a6' for k in k_values]

bars = plt.barh(strategies, k_values, color=colors, edgecolor='#555555', height=0.6)
plt.axvline(x=0, color='#333333', linestyle='-', linewidth=1.2)
plt.xlabel("Causal Support K(E : F) in Bits (Log-Likelihood Ratio)", fontsize=12, fontweight='bold')
plt.title("I.J. Good: Causal Weight of Evidence for Cybernetic Strategies", fontsize=13, fontweight='bold', pad=15)
plt.grid(True, linestyle=':', alpha=0.5)

# Add values to the ends of bars
for bar in bars:
    width = bar.get_width()
    label_x = width + (0.02 if width >= 0 else -0.09)
    plt.text(label_x, bar.get_y() + bar.get_height()/2, f"{width:+.4f} bits",
             va='center', ha='left' if width >= 0 else 'right', fontweight='bold', fontsize=10,
             color='#2c3e50')

plt.tight_layout()
plt.show()
```
