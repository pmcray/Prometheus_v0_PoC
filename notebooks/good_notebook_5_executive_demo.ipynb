## 🎴 Part 1: Go Implementation

Demonstrate complete Go game with Prometheus agents.

```python
# Create Go environment (9x9 for faster demo)
go_env = GoEnvironment(board_size=9, komi=7.5)

print("🎴 Go Environment Created")
print(f"   Board size: {go_env.board.size}×{go_env.board.size}")
print(f"   Komi: {go_env.board.komi}")
print(f"
   Initial state shape: {go_env.get_state().shape}")
print(f"   Legal moves: {len(go_env.get_legal_moves())} positions + pass")
```

```python
# Create Go agents
print("Creating Go agents...")

# Random baseline
random_go = RandomGoAgent(board_size=9)
print(f"✅ {random_go.name} created")

# Prometheus Go agent (trainable)
prometheus_go = PrometheusGoAgent(
    board_size=9,
    num_input_planes=3,
    online_lr=0.0001
)
print(f"✅ {prometheus_go.name} created")
```

```python
# Demo: Play match between agents
print("
" + "="*60)
print("🥊 DEMO MATCH: Random vs Prometheus (before training)")
print("="*60)

match_results = play_match(
    agent1=prometheus_go,
    agent2=random_go,
    num_games=3,
    board_size=9,
    verbose=True
)

print(f"
📊 Match Results:")
print(f"   Prometheus: {match_results['agent1_wins']} wins")
print(f"   Random: {match_results['agent2_wins']} wins")
print(f"   Draws: {match_results['draws']}")
```

## 💾 Part 4: Model Persistence & Checkpointing

Save and load trained models for quick demos.

```python
# Create checkpoint manager
checkpoint_mgr = ModelCheckpoint(checkpoint_dir="./checkpoints")

print("💾 Checkpoint Manager Created")
print(f"   Directory: {checkpoint_mgr.checkpoint_dir}")
print()

# Save current Go agent
checkpoint_mgr.save_agent(
    agent=prometheus_go,
    name="prometheus_go_demo",
    metadata={
        'board_size': 9,
        'architecture': 'policy_value_network',
        'description': 'Demo Go agent from executive notebook'
    }
)

print("✅ Agent saved!")
print()
print("To load later:")
print("  loaded_agent = checkpoint_mgr.load_agent(")
print("      name='prometheus_go_demo',")
print("      agent_class=PrometheusGoAgent,")
print("      board_size=9")
print("  )")
```

## 🔄 Part 5: Self-Play Training (Recursive Self-Improvement)

Demonstrate Prometheus improving through self-play.

```python
# Quick self-play training demo
from prometheus.training.go_training import train_agent_selfplay

print("🚀 Self-Play Training Demo")
print("="*60)
print()

# Train for a few iterations (increase for real training)
history = train_agent_selfplay(
    agent=prometheus_go,
    num_iterations=3,
    games_per_iteration=5,
    epochs_per_iteration=2,
    board_size=9,
    verbose=True
)

print()
print("="*60)
print("✅ Training complete!")
print()

# Plot training progress
if history['policy_loss']:
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    
    axes[0].plot(history['iteration'], history['policy_loss'], marker='o')
    axes[0].set_xlabel('Iteration')
    axes[0].set_ylabel('Policy Loss')
    axes[0].set_title('Policy Network Training')
    axes[0].grid(True, alpha=0.3)
    
    axes[1].plot(history['iteration'], history['value_loss'], marker='o', color='orange')
    axes[1].set_xlabel('Iteration')
    axes[1].set_ylabel('Value Loss')
    axes[1].set_title('Value Network Training')
    axes[1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
```
