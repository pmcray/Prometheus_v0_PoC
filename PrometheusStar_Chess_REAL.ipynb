## Chess AI Opponents

These are real chess AI - they follow actual chess rules and make legal moves.

```python
def evaluate_board(board: chess.Board) -> float:
    """Evaluate chess position (material + position)"""
    if board.is_checkmate():
        return -10000 if board.turn else 10000
    if board.is_stalemate() or board.is_insufficient_material():
        return 0
    
    # Piece values
    piece_values = {
        chess.PAWN: 1,
        chess.KNIGHT: 3,
        chess.BISHOP: 3,
        chess.ROOK: 5,
        chess.QUEEN: 9,
        chess.KING: 0
    }
    
    score = 0
    for square in chess.SQUARES:
        piece = board.piece_at(square)
        if piece:
            value = piece_values[piece.piece_type]
            score += value if piece.color == chess.WHITE else -value
    
    return score

class ChessAI:
    """Real chess AI opponents"""
    
    @staticmethod
    def random_move(board: chess.Board) -> chess.Move:
        """Pick random legal move"""
        return random.choice(list(board.legal_moves))
    
    @staticmethod
    def greedy_move(board: chess.Board) -> chess.Move:
        """Pick move with best immediate material gain"""
        best_move = None
        best_score = float('-inf')
        
        for move in board.legal_moves:
            board.push(move)
            score = evaluate_board(board)
            board.pop()
            
            if score > best_score:
                best_score = score
                best_move = move
        
        return best_move or ChessAI.random_move(board)
    
    @staticmethod
    def minimax_move(board: chess.Board, depth: int = 2) -> chess.Move:
        """Minimax search (looks ahead)"""
        def minimax(board, depth, maximizing):
            if depth == 0 or board.is_game_over():
                return evaluate_board(board)
            
            if maximizing:
                max_eval = float('-inf')
                for move in board.legal_moves:
                    board.push(move)
                    eval_score = minimax(board, depth - 1, False)
                    board.pop()
                    max_eval = max(max_eval, eval_score)
                return max_eval
            else:
                min_eval = float('inf')
                for move in board.legal_moves:
                    board.push(move)
                    eval_score = minimax(board, depth - 1, True)
                    board.pop()
                    min_eval = min(min_eval, eval_score)
                return min_eval
        
        best_move = None
        best_score = float('-inf')
        
        for move in board.legal_moves:
            board.push(move)
            score = minimax(board, depth - 1, False)
            board.pop()
            
            if score > best_score:
                best_score = score
                best_move = move
        
        return best_move or ChessAI.random_move(board)

print("✅ Real chess AI opponents ready!")
print("   - Random: Picks any legal move")
print("   - Greedy: Maximizes material")
print("   - Minimax: Plans ahead (depth 2)")
```

## Play Real Chess Game

Let's watch a real game between two AIs:

```python
def play_chess_game(white_ai, black_ai, max_moves: int = 100, verbose: bool = True) -> Dict:
    """Play a real chess game"""
    board = chess.Board()
    moves = []
    
    if verbose:
        print("\n♟️  Starting chess game...")
    
    while not board.is_game_over() and len(moves) < max_moves:
        # Current player's move
        if board.turn == chess.WHITE:
            move = white_ai(board)
        else:
            move = black_ai(board)
        
        board.push(move)
        moves.append(move.uci())
        
        if verbose and len(moves) % 10 == 0:
            print(f"  Move {len(moves)}: {move.uci()}")
    
    # Determine result
    if board.is_checkmate():
        winner = "Black" if board.turn == chess.WHITE else "White"
        result = "checkmate"
    elif board.is_stalemate():
        winner = "Draw"
        result = "stalemate"
    elif board.is_insufficient_material():
        winner = "Draw"
        result = "insufficient_material"
    else:
        winner = "Draw"
        result = "max_moves"
    
    if verbose:
        print(f"\n  Game over: {winner} ({result})")
        print(f"  Total moves: {len(moves)}")
    
    return {
        'winner': winner,
        'result': result,
        'moves': len(moves),
        'pgn': ' '.join(moves)
    }

# Test with real game
print("Random vs Greedy:")
result = play_chess_game(ChessAI.random_move, ChessAI.greedy_move, verbose=True)

print("\n" + "="*50)
print("✅ That was a REAL chess game!")
print(f"Winner: {result['winner']}")
print(f"Moves: {result['moves']}")
print("="*50)
```

## Evolve Chess Strategy

Now let's evolve a strategy that plays real chess:

```python
class ChessStrategyAI:
    """Parameterized chess AI that can be evolved"""
    
    def __init__(self, params: Dict):
        self.params = params
    
    def get_move(self, board: chess.Board) -> chess.Move:
        """Pick move based on evolved parameters"""
        best_move = None
        best_score = float('-inf')
        
        for move in board.legal_moves:
            board.push(move)
            
            # Evaluate with evolved weights
            material = evaluate_board(board)
            mobility = board.legal_moves.count()
            
            score = (
                material * self.params['material_weight'] +
                mobility * self.params['mobility_weight']
            )
            
            # Capture bonus
            if board.is_capture(move):
                score += self.params['capture_bonus']
            
            board.pop()
            
            if score > best_score:
                best_score = score
                best_move = move
        
        return best_move or ChessAI.random_move(board)

def random_params() -> Dict:
    return {
        'material_weight': random.uniform(0.5, 2.0),
        'mobility_weight': random.uniform(0.0, 0.5),
        'capture_bonus': random.uniform(0, 50)
    }

print("✅ Evolvable chess strategy ready!")
```

## Run Chess Curriculum

Evolve strategies against progressively harder opponents:

```python
# Chess Curriculum
CURRICULUM = [
    {'stage': 1, 'name': 'Random', 'opponent': ChessAI.random_move, 'generations': 20},
    {'stage': 2, 'name': 'Greedy', 'opponent': ChessAI.greedy_move, 'generations': 30},
    {'stage': 3, 'name': 'Minimax', 'opponent': ChessAI.minimax_move, 'generations': 40},
]

print("="*70)
print("♟️  PROMETHEUSSTAR CHESS CURRICULUM (REAL GAMES!)")
print("="*70)
print()

# Initialize population
population_size = 10
population = [random_params() for _ in range(population_size)]
GAMES_PER_EVAL = 3  # More games for better statistics

for stage in CURRICULUM:
    print(f"\n{'='*70}")
    print(f"STAGE {stage['stage']}: vs {stage['name']} AI")
    print(f"Generations: {stage['generations']}")
    print(f"Population: {population_size} agents")
    print(f"Games per agent: {GAMES_PER_EVAL}")
    print(f"Total games per generation: {population_size * GAMES_PER_EVAL}")
    print('='*70)
    
    start_time = time.time()
    
    for gen in range(stage['generations']):
        # Evaluate population - collect detailed stats
        fitness_scores = []
        all_results = []
        
        for agent_id, params in enumerate(population):
            ai = ChessStrategyAI(params)
            agent_wins = 0
            agent_draws = 0
            agent_losses = 0
            agent_results = []
            
            # Play multiple games per evaluation
            for game_num in range(GAMES_PER_EVAL):
                result = play_chess_game(
                    ai.get_move,
                    stage['opponent'],
                    max_moves=50,
                    verbose=False
                )
                
                if result['winner'] == 'White':
                    agent_wins += 1
                elif result['winner'] == 'Draw':
                    agent_draws += 1
                else:
                    agent_losses += 1
                
                agent_results.append(result)
            
            all_results.append({
                'agent_id': agent_id,
                'wins': agent_wins,
                'draws': agent_draws,
                'losses': agent_losses,
                'games': agent_results,
                'params': params
            })
            
            # Fitness = wins * 1000 + draws * 100
            fitness_scores.append(agent_wins * 1000 + agent_draws * 100 + random.randint(0, 10))
        
        # Get best agent
        best_idx = fitness_scores.index(max(fitness_scores))
        best_agent = all_results[best_idx]
        
        # Calculate generation statistics
        total_wins = sum(r['wins'] for r in all_results)
        total_draws = sum(r['draws'] for r in all_results)
        total_losses = sum(r['losses'] for r in all_results)
        total_games = population_size * GAMES_PER_EVAL
        avg_moves = sum(g['moves'] for r in all_results for g in r['games']) / total_games
        
        # Progress update
        if gen % 5 == 0 or gen == stage['generations'] - 1:
            elapsed = time.time() - start_time
            eta = (elapsed / (gen + 1)) * (stage['generations'] - gen - 1)
            
            print(f"\n  Generation {gen+1}/{stage['generations']}:")
            print(f"  ├─ Games played: {total_games}")
            print(f"  ├─ Overall: W:{total_wins} D:{total_draws} L:{total_losses} ({total_wins/total_games:.0%} win rate)")
            print(f"  ├─ Avg moves/game: {avg_moves:.1f}")
            print(f"  ├─ Best agent: W:{best_agent['wins']} D:{best_agent['draws']} L:{best_agent['losses']} ({best_agent['wins']/GAMES_PER_EVAL:.0%})")
            print(f"  └─ ETA: {eta/60:.1f}m")
            
            # Show winning position for best agent with GRAPHICAL BOARD
            for game in best_agent['games']:
                if game['winner'] == 'White':
                    # Replay game to final position
                    board = chess.Board()
                    moves = game['pgn'].split()
                    for move_str in moves:
                        board.push(chess.Move.from_uci(move_str))
                    
                    print(f"\n  🏆 Winning position ({game['result']}, {game['moves']} moves):")
                    display(SVG(chess.svg.board(board, size=300)))
                    break  # Show only first win
        
        # Evolution (keep best, mutate rest)
        new_population = [best_agent['params']]
        
        while len(new_population) < population_size:
            parent = best_agent['params']
            child = {
                'material_weight': max(0.5, min(2.0, parent['material_weight'] + random.gauss(0, 0.2))),
                'mobility_weight': max(0.0, min(0.5, parent['mobility_weight'] + random.gauss(0, 0.1))),
                'capture_bonus': max(0, min(100, parent['capture_bonus'] + random.gauss(0, 10)))
            }
            new_population.append(child)
        
        population = new_population
    
    elapsed = time.time() - start_time
    print(f"\n  ✅ Stage {stage['stage']} complete in {elapsed/60:.1f} minutes")
    print(f"  Final best params: material={best_agent['params']['material_weight']:.2f}, " +
          f"mobility={best_agent['params']['mobility_weight']:.2f}, " +
          f"capture={best_agent['params']['capture_bonus']:.1f}")

print("\n" + "="*70)
print("🎉 CHESS CURRICULUM COMPLETE!")
print("="*70)
print("\nEvolved strategy that can beat Random, Greedy, and Minimax!")
print("This was 100% REAL CHESS - no mocking! ♟️")
```
